# 2 -  Create Useful Features

2.1 Price Change

```python
# Create price change
df_prices['price_change'] = (
df_prices.groupby('ticker', observed=True)['close'].diff()
)
```

2.2 Daily Return

```python
# Create daily return
df_prices['daily_return'] = (
df_prices.groupby('ticker', observed=True)['close'].pct_change(fill_method=None)
)
```

2.3 Previous-Day Closing Price

```python
# Create previous-day closing price
df_prices['close_lag_1'] = (
df_prices.groupby('ticker', observed=True)['close'].shift(1)
)
```

2.4 Moving Averages

```python
# Create 5-day moving average
df_prices['ma_5'] = (
df_prices.groupby('ticker', observed=True)['close']
.transform(lambda x: x.rolling(window=5).mean())
)
```

```python
# Create 20-day moving average
df_prices['ma_20'] = (
df_prices.groupby('ticker', observed=True)['close']
.transform(lambda x: x.rolling(window=20).mean())
)
```

2.5 Volatility

```python
# Create 20-day volatility
df_prices['volatility_20'] = (
df_prices.groupby('ticker', observed=True)['daily_return']
.transform(lambda x: x.rolling(window=20).std())
)
```

2.6 Additional Feature: Daily High-Low Range

```python
# Create percentage high-low range
df_prices['high_low_range_pct'] = (
(df_prices['high'] - df_prices['low']) / df_prices['low']
)
```

Select the engineered features that require previous observations

3.3 Encode Categorical Variables

```python
# Check the number of categories
print(df_merged[['exchange', 'sector','industry']].nunique())

```

```python
# Replace missing sector values with an explicit 'Unknown' category
df_merged['sector'] = df_merged['sector'].fillna('Unknown')
```

```python
# Remove industry to reduce the number of categorical variables
df_merged.drop(columns=['industry'], inplace=True)
```

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Create the encoder
encoder = OneHotEncoder(
sparse_output=True,
dtype=np.int8,
handle_unknown='ignore'
)
```

```python
# Encode exchange and sector
encoded_features = encoder.fit_transform(
df_merged[['exchange', 'sector']]
)

# Check the result
print("Encoded shape:", encoded_features.shape)
```

```python
# Display encoded feature names
encoded_names = encoder.get_feature_names_out(
['exchange', 'sector']
)

print(encoded_names)
```
